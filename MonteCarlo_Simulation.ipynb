```python
cards = 980
trials = []
price = 25

def collect(cards, threshold):

    packs = 0
    deck = set()
    duplicates = {}



    while len(deck) < cards * threshold:

        packs += 1




        for i in range(5):
            card = random.randint(1, cards)
            if card not in deck:
              deck.add(card)
            else:
              duplicates[card] = duplicates.get(card, 0) + 1


    missing = cards - len(deck)
    extra = sum(duplicates.values())
    trades = min(extra, missing)
    completed = len(deck) + trades


    return packs, completed


thresholds = []
probs = []
avg_packs = []

for threshold in [x/1000 for x in range(620, 661)]:
  success = 0
  total_packs = 0
  for i in range(5000):
    packs, completed = collect(cards, threshold)
    total_packs += packs
    if completed >= cards:
      success += 1


  thresholds.append(threshold)
  probs.append(success/5000)
  avg_packs.append(total_packs/5000)
```

```python
fig, ax1 = plt.subplots()

ax1.plot(thresholds, probs, label='Probability')

ax2 = ax1.twinx()
ax2.plot(thresholds, avg_packs, color='red', label='Average Packs')

h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()

ax1.legend(h1 + h2, l1 + l2, loc='lower right')

plt.title('Probability of Completion vs. Average Packs Needed')

plt.show()
```
